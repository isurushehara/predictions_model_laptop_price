```python
data['SSD'] = 0
data['HDD'] = 0
data['Hybrid'] = 0
data['Flash_Storage'] = 0
```

```python
import re

def process_memory(memory):

    parts = memory.split("+")

    ssd = 0
    hdd = 0
    hybrid = 0
    flash = 0

    for part in parts:

        part = part.strip()

        # Extract the numeric value
        size = re.findall(r'(\d+\.?\d*)', part)

        if len(size) == 0:
            continue

        size = float(size[0])

        # Convert TB to GB
        if size <= 2:
            size *= 1024

        if 'SSD' in part:
            ssd += size

        elif 'HDD' in part:
            hdd += size

        elif 'Hybrid' in part:
            hybrid += size

        elif 'Flash Storage' in part:
            flash += size

    return pd.Series([ssd, hdd, hybrid, flash])
```

```python
data[['SSD','HDD','Hybrid','Flash_Storage']] = data['Memory'].apply(process_memory)
```

```python
def model_acc(model):
    model.fit(X_train, y_train)
    acc = model.score(X_test, y_test)
    print(str(model)+ '--> ' +str(acc))
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
model_acc(lr)

from sklearn.linear_model import Lasso
lasso = Lasso()
model_acc(lasso)

from sklearn.tree import DecisionTreeRegressor
lasso = Lasso()
model_acc(lasso)
```
